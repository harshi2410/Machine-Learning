```python
documents = [
    "Artificial Intelligence is the simulation of human intelligence.",
    "Machine Learning is a subset of AI that learns from data.",
    "Deep Learning uses neural networks with many layers.",
    "Python is commonly used in AI development."
]
```

```python
chunks = documents  # already small, so no need to split
```

4: Create Embeddings   Converts text → vectors (numbers)
Similar meaning → similar vectors

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer('all-MiniLM-L6-v2')

embeddings = model.encode(chunks)
```

```python
import faiss
import numpy as np

dimension = embeddings.shape[1]

index = faiss.IndexFlatL2(dimension)
index.add(np.array(embeddings))
```

```python
query = "What is AI?"

query_embedding = model.encode([query])

k = 2  # top results
distances, indices = index.search(query_embedding, k)

retrieved_docs = [chunks[i] for i in indices[0]]

print("Retrieved:", retrieved_docs)
```

```python
context = " ".join(retrieved_docs)

final_answer = f"""
Answer based on context:

Context: {context}

Question: {query}

Answer: {context}
"""

print(final_answer)
```

```python
from transformers import pipeline

generator = pipeline("text-generation", model="distilgpt2")

prompt = f"""
Answer based only on the context below:

Context:
{context}

Question:
{query}

Answer:
"""

response = generator(prompt, max_length=100)

print(response[0]['generated_text'])
```

```python
documents = [
    "Artificial Intelligence enables machines to mimic human intelligence.",
    "Machine Learning allows systems to learn from data without explicit programming.",
    "Deep Learning uses neural networks with many hidden layers.",
    "Natural Language Processing helps machines understand human language.",
    "Python is widely used for AI, ML, and data science applications."
]
```

```python
def chunk_text(text, chunk_size=50):
    words = text.split()
    chunks = []
    for i in range(0, len(words), chunk_size):
        chunks.append(" ".join(words[i:i+chunk_size]))
    return chunks

chunks = []
for doc in documents:
    chunks.extend(chunk_text(doc))
```

```python
from sentence_transformers import SentenceTransformer

embed_model = SentenceTransformer('all-MiniLM-L6-v2')
embeddings = embed_model.encode(chunks)
```

```python
import faiss
import numpy as np

dimension = embeddings.shape[1]
index = faiss.IndexFlatL2(dimension)
index.add(np.array(embeddings))
```

```python
def retrieve(query, k=2):
    query_embedding = embed_model.encode([query])
    distances, indices = index.search(query_embedding, k)
    return [chunks[i] for i in indices[0]]
```

```python
from transformers import pipeline

generator = pipeline("text-generation", model="distilgpt2")
```

```python
def rag_pipeline(query):
    retrieved_docs = retrieve(query)
    context = " ".join(retrieved_docs)

    prompt = f"""
You are a helpful AI assistant.
Answer ONLY from the provided context.
If answer is not in context, say "I don't know".

Context:
{context}

Question:
{query}

Answer:
"""

    response = generator(prompt, max_length=120)
    return response[0]['generated_text']
```

```python
while True:
    q = input("Ask: ")
    if q.lower() == "exit":
        break

    answer = rag_pipeline(q)
    print("\nAI:", answer)
```
