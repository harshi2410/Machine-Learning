```python
import gradio as gr
from transformers import pipeline
```

```python
classifier = pipeline("sentiment-analysis")
```

```python
def analyze_text(text):
    result = classifier(text)[0]

    if result['label'] == 'POSITIVE':
        emoji = "😊"
    else:
        emoji = "😠"

    return f"{emoji} {result['label']} (Confidence: {result['score']:.2f})"
```

```python
interface = gr.Interface(
    fn=analyze_text,
    inputs=gr.Textbox(lines=3, placeholder="Enter your text here..."),
    outputs="text",
    title="AI Text Analyzer 🚀",
    description="Analyze sentiment using Transformer model"
)
```

```python
interface.launch()
```
